```python
def odd_even(num):

  if num % 2==0:
    return '짝수'
  else:
    return '홀수'

int_num=int(input('정수를 입력:'))
print(odd_even(int_num))
```

```python
def triangle_area(base, height):
    area = (base * height) / 2
    return area

# 밑변과 높이 입력받기
base = float(input("삼각형의 밑변을 입력하세요: "))
height = float(input("삼각형의 높이를 입력하세요: "))

# 넓이 계산 및 출력
result = triangle_area(base, height)
print("삼각형의 넓이:", result)
```
