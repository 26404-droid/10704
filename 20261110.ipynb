```python
def calculator(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            return "0으로 나눌 수 없습니다."
        return a / b
    else:
        return "잘못된 연산자입니다."


# 사용자 입력
a = float(input("첫 번째 숫자: "))
operator = input("연산자 (+, -, *, /): ")
b = float(input("두 번째 숫자: "))

result = calculator(a, b, operator)

print("계산 결과:", result)
```
