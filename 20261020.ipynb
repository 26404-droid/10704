```python
director=input('감독은 누구')
stadium=input('경기장은 어디')
win=input('이긴 팀 이름')
loser=input('진 팀 이름')
score=input('몇대몇')
vip=input('우승선수')
print('------주요 뉴스-------')
print('오늘 %s에서 축구 경기가 열렸습니다.'% stadium)
print('%s 감독이 이끄는 %s은 %s를 상대로 치열한 경기를 하였습니다'%(director,win,loser))
print('그 결과 %s(으)로 승리를 거거두었고 우승 선수는 %s 선수입니다.'%(score,vip))
```
